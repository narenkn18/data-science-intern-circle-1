```python
#Question 1:Did gender influence the survival rate of Titanic passengers?
#Question 2:How did passenger class (Pclass) affect survival chances?
#Question 3:Did the port of embarkation influence passenger survival rates?
import pandas as pd
import seaborn as sns
df=sns.load_dataset("titanic")
print("Dataset loaded successfully")
print()
print(df.shape)
print("Dataset contains 891 rows and 15 columns")
print()
print(df.head())
print("Displays first 5 rows")
print("""This datset contains 
1.Survived,
2.sex(Gender of passenger),
3.age,
4.pclass(passenger ticket class),
5.parch(no of parents or childeren trvaleling with passenger,
6.Fare(ticket price),
7.embarked(where passenger boarded the titanic,
8.class(ticket class),
9.adult male(is passenger adult male True or False,
10.embark town(boarding place),
11.alive(survived or not),
12.alone(Travelling alone or not)""") 
print()
```

```python
print("Last 5 rows :",df.tail())
print()
print("info:",df.info())
print()
print("Describe:",df.describe())
print()
```

```python
print("Missing Values:")
print(df.isnull())
print("No.of missing values:")
print(df.isnull().sum())
print("No of Notnull values:")
print(df.notnull().sum())
print("Filling null values in Age column with median of Age column")
df["age"].fillna(df["age"].median(),inplace=True)
print("No. of null values in Age column after filling median values:",df["age"].isnull().sum())
print("77% of data in Deck column is empty so its useless,we will remove that column")
df.drop("deck",axis=1,inplace=True)
print("columns after removing Deck column:")
print(df.columns)
print("Filling embark column with most repeated town name")
df["embark_town"].fillna(df["embark_town"].mode()[0],inplace=True)
df["embarked"].fillna(df["embarked"].mode()[0],inplace=True)
print("null values after filling embark column")
print(df.isnull().sum())
```

```python
print("No of duplicated values in the dataset:")
print(df.duplicated().sum())
print("Duplicated rows are:")
print(df[df.duplicated()])
duplicate_count = df.duplicated().sum()
print("No of duplicated values in the dataset:")
print(duplicate_count)
print("\nDuplicated rows are:")
print(df[df.duplicated()])
```

```python
#Survival Rate by gender
survival_by_sex = df.groupby("sex")["survived"].mean() * 100
print(survival_by_sex.sort_values(ascending=False))
print()
#Survival rate by passanger class
survival_by_class = df.groupby("pclass")["survived"].mean() * 100
print(survival_by_class.sort_values(ascending=False))
print()
#survival rate by age
bins = [0, 12, 18, 35, 60, 100]
labels = ["Child", "Teenager", "Young Adult", "Adult", "Senior"]
df["age_band"] = pd.cut(df["age"], bins=bins, labels=labels)
survival_by_age = df.groupby("age_band", observed=False)["survived"].mean() * 100
print(survival_by_age.sort_values(ascending=False))
print()
#Survival rate by gender and class
survival_by_sex_class = df.groupby(["sex", "pclass"])["survived"].mean() * 100
print(survival_by_sex_class.sort_values(ascending=False))
print()
#Survival by port of embarkon
df.groupby("embark_town")["pclass"].value_counts()
```
